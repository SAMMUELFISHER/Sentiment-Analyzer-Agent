```python
def reverse_string_loop(text):
    reversed_text = ""
    for char in text:
        # Prepend the character to the front
        reversed_text = char + reversed_text 
    return reversed_text

# Example Usage:
original = "hello"
print(reverse_string_loop(original))  # Output: "olleh"
```
